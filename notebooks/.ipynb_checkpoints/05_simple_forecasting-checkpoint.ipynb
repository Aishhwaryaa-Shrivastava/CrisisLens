# 🔮 Week 3 Day 15-17: Simple Forecasting Models

**Goal:** Build baseline forecasting models to predict risk 1-3 months ahead

**What You'll Build:**
1. Moving Average Model (uses last 3 months average)
2. Linear Trend Model (extends trend line forward)
3. Comparison & Validation

**Deliverables:**
- Forecasts for Feb-Apr 2026 (3 months ahead)
- Model accuracy comparison
- Baseline to compare Prophet against later

## 📦 Setup & Data Loading

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from datetime import datetime, timedelta
from dateutil.relativedelta import relativedelta
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('default')
sns.set_palette('husl')

print("Libraries loaded successfully!")
```

```python
# Load data
data = pd.read_csv('./data/processed/combined_monthly_2024_2025.csv')

# Convert month to datetime
data['month_date'] = pd.to_datetime(data['month'], format='%Y-%m')

# Sort by state and date
data = data.sort_values(['state', 'month_date']).reset_index(drop=True)

print(f"Loaded Data:")
print(f"   Total records: {len(data):,}")
print(f"   Date range: {data['month'].min()} to {data['month'].max()}")
print(f"   Regions: {data['state'].nunique()}")

# Display sample
data.head()
```

## 🎯 Part 1: Train-Test Split

We'll use:
- **Training data**: Jan 2024 - Oct 2025 (22 months)
- **Validation data**: Nov-Dec 2025 (2 months) - to test accuracy
- **Forecast target**: Feb-Apr 2026 (3 months ahead)

```python
# Define split dates
train_end = pd.to_datetime('2025-10-01')
validation_start = pd.to_datetime('2025-11-01')
validation_end = pd.to_datetime('2025-12-01')

# Split data
train_data = data[data['month_date'] <= train_end].copy()
validation_data = data[data['month_date'] > train_end].copy()

print("Data Split:")
print(f"   Training: {len(train_data)} records ({train_data['month'].min()} to {train_data['month'].max()})")
print(f"   Validation: {len(validation_data)} records ({validation_data['month'].min()} to {validation_data['month'].max()})")
print(f"\n   Will forecast: Feb 2026, Mar 2026, Apr 2026")
```

## 📊 Part 2: Model 1 - Moving Average Forecasting

```python
def moving_average_forecast(state_name, train_df, months_ahead=1, window=3):
    """
    Predict future risk using moving average of last N months
    
    Args:
        state_name: State to forecast
        train_df: Training data
        months_ahead: How many months to forecast
        window: Number of months to average (default 3)
    
    Returns:
        Predicted risk value
    """
    # Get state data
    state_data = train_df[train_df['state'] == state_name].copy()
    
    # Get last N months
    recent_data = state_data.tail(window)
    
    # Calculate average
    forecast = recent_data['total_risk'].mean()
    
    return forecast

# Test on one state
test_state = 'Maharashtra'
ma_prediction = moving_average_forecast(test_state, train_data, months_ahead=1, window=3)

print(f"Moving Average Test - {test_state}:")
print(f"   Last 3 months average risk: {ma_prediction:.2f}")
print(f"   Prediction for Jan 2026: {ma_prediction:.2f}")
```

## 📈 Part 3: Model 2 - Linear Trend Forecasting

```python
def linear_trend_forecast(state_name, train_df, months_ahead=1):
    """
    Predict future risk by extending linear trend line
    
    Args:
        state_name: State to forecast
        train_df: Training data
        months_ahead: How many months to forecast
    
    Returns:
        Predicted risk value
    """
    # Get state data
    state_data = train_df[train_df['state'] == state_name].copy()
    
    # Prepare regression data
    x = np.arange(len(state_data))  # 0, 1, 2, ..., n
    y = state_data['total_risk'].values
    
    # Linear regression
    slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
    
    # Predict future point
    future_x = len(state_data) + months_ahead - 1
    forecast = slope * future_x + intercept
    
    return forecast

# Test on same state
lt_prediction = linear_trend_forecast(test_state, train_data, months_ahead=1)

print(f"Linear Trend Test - {test_state}:")
print(f"   Prediction for Jan 2026: {lt_prediction:.2f}")
```

## 🔍 Part 4: Validate Models on Nov-Dec 2025 Data

```python
# Validate both models
validation_results = []

for state in data['state'].unique():
    # Get actual values for Nov and Dec 2025
    state_validation = validation_data[validation_data['state'] == state]
    
    if len(state_validation) == 0:
        continue
    
    # Actual values
    actual_nov = state_validation[state_validation['month'] == '2025-11']['total_risk'].values
    actual_dec = state_validation[state_validation['month'] == '2025-12']['total_risk'].values
    
    if len(actual_nov) == 0 or len(actual_dec) == 0:
        continue
    
    actual_nov = actual_nov[0]
    actual_dec = actual_dec[0]
    
    # Predictions
    ma_pred_nov = moving_average_forecast(state, train_data, months_ahead=1, window=3)
    ma_pred_dec = moving_average_forecast(state, train_data, months_ahead=2, window=3)
    
    lt_pred_nov = linear_trend_forecast(state, train_data, months_ahead=1)
    lt_pred_dec = linear_trend_forecast(state, train_data, months_ahead=2)
    
    # Calculate errors
    ma_error_nov = abs(ma_pred_nov - actual_nov)
    ma_error_dec = abs(ma_pred_dec - actual_dec)
    
    lt_error_nov = abs(lt_pred_nov - actual_nov)
    lt_error_dec = abs(lt_pred_dec - actual_dec)
    
    validation_results.append({
        'state': state,
        'actual_nov': actual_nov,
        'actual_dec': actual_dec,
        'ma_pred_nov': ma_pred_nov,
        'ma_pred_dec': ma_pred_dec,
        'lt_pred_nov': lt_pred_nov,
        'lt_pred_dec': lt_pred_dec,
        'ma_error_avg': (ma_error_nov + ma_error_dec) / 2,
        'lt_error_avg': (lt_error_nov + lt_error_dec) / 2
    })

validation_df = pd.DataFrame(validation_results)

print("Validation Complete!")
print(f"\nModel Performance (Average Absolute Error):")
print(f"   Moving Average Model: {validation_df['ma_error_avg'].mean():.2f}")
print(f"   Linear Trend Model: {validation_df['lt_error_avg'].mean():.2f}")

# Determine winner
if validation_df['ma_error_avg'].mean() < validation_df['lt_error_avg'].mean():
    print(f"\n   WINNER: Moving Average (lower error)")
    best_model = 'Moving Average'
else:
    print(f"\n   WINNER: Linear Trend (lower error)")
    best_model = 'Linear Trend'

validation_df.head(10)
```

## 🎯 Part 5: Generate Feb-Apr 2026 Forecasts

```python
# Generate forecasts for all states, 3 months ahead
forecasts_2026 = []

for state in data['state'].unique():
    # Get current (Dec 2025) risk
    state_data = data[data['state'] == state]
    current_risk = state_data['total_risk'].iloc[-1]
    
    # Forecast Feb, Mar, Apr 2026
    ma_feb = moving_average_forecast(state, data, months_ahead=2, window=3)
    ma_mar = moving_average_forecast(state, data, months_ahead=3, window=3)
    ma_apr = moving_average_forecast(state, data, months_ahead=4, window=3)
    
    lt_feb = linear_trend_forecast(state, data, months_ahead=2)
    lt_mar = linear_trend_forecast(state, data, months_ahead=3)
    lt_apr = linear_trend_forecast(state, data, months_ahead=4)
    
    forecasts_2026.append({
        'state': state,
        'current_risk_dec2025': current_risk,
        'ma_forecast_feb2026': ma_feb,
        'ma_forecast_mar2026': ma_mar,
        'ma_forecast_apr2026': ma_apr,
        'lt_forecast_feb2026': lt_feb,
        'lt_forecast_mar2026': lt_mar,
        'lt_forecast_apr2026': lt_apr
    })

forecasts_df = pd.DataFrame(forecasts_2026)

# Add best model forecast
if best_model == 'Moving Average':
    forecasts_df['best_forecast_feb'] = forecasts_df['ma_forecast_feb2026']
    forecasts_df['best_forecast_mar'] = forecasts_df['ma_forecast_mar2026']
    forecasts_df['best_forecast_apr'] = forecasts_df['ma_forecast_apr2026']
else:
    forecasts_df['best_forecast_feb'] = forecasts_df['lt_forecast_feb2026']
    forecasts_df['best_forecast_mar'] = forecasts_df['lt_forecast_mar2026']
    forecasts_df['best_forecast_apr'] = forecasts_df['lt_forecast_apr2026']

print("Feb-Apr 2026 Forecasts Generated!")
print(f"\nTop 10 High-Risk States (Feb 2026 - {best_model} Model):")
print(forecasts_df.nlargest(10, 'best_forecast_feb')[['state', 'current_risk_dec2025', 'best_forecast_feb']])
```
