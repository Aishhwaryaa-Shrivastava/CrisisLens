```python
import pandas as pd

# Load the merged data
data = pd.read_csv('data/processed/combined_monthly_2024_2025.csv')

print("=" * 60)
print("✅ DATA VERIFICATION")
print("=" * 60)
print(f"\nTotal months: {len(data)}")
print(f"Regions: {data['state'].nunique()}")
print(f"Date range: {data['month'].min()} to {data['month'].max()}")
print(f"\nExpected: 864 months (36 regions × 24 months)")
print(f"Missing values: {data.isnull().sum().sum()}")

print("\n✅ Top 5 High-Risk States (Average):")
top_states = data.groupby('state')['total_risk'].mean().nlargest(5)
for state, risk in top_states.items():
    print(f"   {state}: {risk:.2f}")

print("\n✅ Sample Data (First 3 rows):")
print(data.head(3))
```
