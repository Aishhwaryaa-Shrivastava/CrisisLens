```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from prophet import Prophet
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('default')
sns.set_palette('husl')

print("✅ Libraries loaded successfully!")
print(f"Prophet version: {Prophet.__module__}")
```

## 📊 Load Data

```python
# Load monthly data
monthly_data = pd.read_csv('../data/processed/combined_monthly_2024_2025.csv')

# Ensure proper datetime format
monthly_data['month_date'] = pd.to_datetime(monthly_data['month'] + '-01')

print("📊 Data Loaded:")
print(f"   Records: {len(monthly_data)}")
print(f"   Date range: {monthly_data['month'].min()} to {monthly_data['month'].max()}")
print(f"   States: {monthly_data['state'].nunique()}")

# Load simple model results for comparison
simple_forecasts = pd.read_csv('../outputs/reports/simple_forecasts_2026.csv')
simple_validation = pd.read_csv('../outputs/reports/forecast_validation_results.csv')

print("\n✅ Simple model results loaded for comparison")
```

## 🔮 Prophet Model Setup

Prophet requires data in specific format:
- `ds`: Date column
- `y`: Value to forecast

```python
def prepare_prophet_data(df, state):
    """
    Prepare data for Prophet model
    
    Args:
        df: Monthly data
        state: State name
    
    Returns:
        DataFrame with 'ds' and 'y' columns
    """
    state_data = df[df['state'] == state].copy()
    state_data = state_data.sort_values('month_date')
    
    # Prophet requires specific column names
    prophet_df = pd.DataFrame({
        'ds': state_data['month_date'],
        'y': state_data['total_risk']
    })
    
    return prophet_df

# Test on one state
test_state = 'Maharashtra'
test_data = prepare_prophet_data(monthly_data, test_state)

print(f"✅ Prophet data prepared for {test_state}:")
print(test_data.head())
print(f"\nRows: {len(test_data)}")
```

## 🎯 Train Prophet Model (Single State Example)

```python
# Create and train Prophet model
model = Prophet(
    yearly_seasonality=True,  # Capture 12-month cycles
    weekly_seasonality=False,  # Not needed for monthly data
    daily_seasonality=False,   # Not needed for monthly data
    seasonality_mode='multiplicative',  # Better for varying amplitude
    interval_width=0.95  # 95% confidence intervals
)

# Fit the model
print(f"Training Prophet model for {test_state}...")
model.fit(test_data)
print("✅ Model trained!")

# Create future dataframe (3 months ahead: Feb, Mar, Apr 2026)
future = model.make_future_dataframe(periods=3, freq='MS')  # MS = Month Start

print(f"\nForecast dates:")
print(future.tail())
```

```python
# Generate forecast
forecast = model.predict(future)

print("📊 Forecast columns:")
print(forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail())

# Extract Feb-Apr 2026 forecasts
feb_2026 = forecast[forecast['ds'] == '2026-02-01']['yhat'].values[0]
mar_2026 = forecast[forecast['ds'] == '2026-03-01']['yhat'].values[0]
apr_2026 = forecast[forecast['ds'] == '2026-04-01']['yhat'].values[0]

print(f"\n🔮 {test_state} Prophet Forecasts:")
print(f"   Feb 2026: {feb_2026:.2f}")
print(f"   Mar 2026: {mar_2026:.2f}")
print(f"   Apr 2026: {apr_2026:.2f}")
```

## 📈 Visualize Prophet Components

```python
# Plot forecast
fig1 = model.plot(forecast, figsize=(12, 6))
plt.title(f'{test_state}: Prophet Forecast with Uncertainty', fontsize=14, fontweight='bold')
plt.ylabel('Multi-Factor Risk Score')
plt.xlabel('Date')
plt.tight_layout()
plt.savefig('../outputs/figures/week3_prophet_example_forecast.png', dpi=150, bbox_inches='tight')
print("✅ Saved: week3_prophet_example_forecast.png")
plt.show()

# Plot components (trend + seasonality)
fig2 = model.plot_components(forecast, figsize=(12, 8))
plt.tight_layout()
plt.savefig('../outputs/figures/week3_prophet_components.png', dpi=150, bbox_inches='tight')
print("✅ Saved: week3_prophet_components.png")
plt.show()
```

## 🔄 Validation: Prophet vs Simple Models

Test Prophet on Nov-Dec 2025 (same validation set as simple models)

```python
def validate_prophet(df, state, validation_month='2025-12-01'):
    """
    Validate Prophet on held-out data
    
    Args:
        df: Full monthly data
        state: State name
        validation_month: Date to validate against
    
    Returns:
        dict with prediction and error
    """
    # Split into train/validation
    train = df[df['month_date'] < pd.to_datetime(validation_month)]
    validation = df[df['month_date'] == pd.to_datetime(validation_month)]
    
    if len(validation) == 0:
        return None
    
    # Prepare data
    train_data = prepare_prophet_data(train, state)
    
    if len(train_data) < 2:
        return None
    
    # Train model
    model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        seasonality_mode='multiplicative'
    )
    
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        model.fit(train_data)
    
    # Forecast
    future = pd.DataFrame({'ds': [pd.to_datetime(validation_month)]})
    forecast = model.predict(future)
    
    # Get actual value
    actual = validation[validation['state'] == state]['total_risk'].values[0]
    predicted = forecast['yhat'].values[0]
    error = abs(actual - predicted)
    
    return {
        'actual': actual,
        'predicted': predicted,
        'error': error
    }

# Test on one state
result = validate_prophet(monthly_data, test_state)
if result:
    print(f"\n✅ Validation for {test_state}:")
    print(f"   Actual (Dec 2025): {result['actual']:.2f}")
    print(f"   Prophet Predicted: {result['predicted']:.2f}")
    print(f"   Error: {result['error']:.2f}")
```

```python
# Validate on all states
print("🔄 Validating Prophet on all states...")
print("This may take 2-3 minutes...\n")

prophet_validation = []

states = monthly_data['state'].unique()
total_states = len(states)

for idx, state in enumerate(states, 1):
    if idx % 5 == 0:
        print(f"Progress: {idx}/{total_states} states...")
    
    result = validate_prophet(monthly_data, state)
    
    if result:
        prophet_validation.append({
            'state': state,
            'actual': result['actual'],
            'prophet_prediction': result['predicted'],
            'prophet_error': result['error']
        })

prophet_val_df = pd.DataFrame(prophet_validation)

# Calculate RMSE
prophet_rmse = np.sqrt(np.mean(prophet_val_df['prophet_error']**2))

print(f"\n✅ Prophet Validation Complete!")
print(f"   Prophet RMSE: {prophet_rmse:.2f}")

# Compare with simple models
ma_rmse = simple_validation['ma_error'].apply(lambda x: x**2).mean()**0.5
lt_rmse = simple_validation['lt_error'].apply(lambda x: x**2).mean()**0.5

print(f"\n📊 Model Comparison:")
print("="*50)
print(f"   Moving Average RMSE: {ma_rmse:.2f}")
print(f"   Linear Trend RMSE:   {lt_rmse:.2f}")
print(f"   Prophet RMSE:        {prophet_rmse:.2f}")
print("="*50)

best_rmse = min(ma_rmse, lt_rmse, prophet_rmse)
if best_rmse == prophet_rmse:
    improvement = ((lt_rmse - prophet_rmse) / lt_rmse) * 100
    print(f"\n🎯 Prophet is BEST MODEL!")
    print(f"   {improvement:.1f}% improvement over Linear Trend")
elif best_rmse == ma_rmse:
    print(f"\n📊 Moving Average is still best")
else:
    print(f"\n📊 Linear Trend is still best")
```

## 🔮 Generate Prophet Forecasts for All States

```python
def forecast_prophet(df, state, periods=3):
    """
    Generate Prophet forecast for a state
    
    Args:
        df: Monthly data
        state: State name
        periods: Months to forecast ahead
    
    Returns:
        dict with forecasts and confidence intervals
    """
    # Prepare data
    prophet_data = prepare_prophet_data(df, state)
    
    # Train model
    model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False,
        seasonality_mode='multiplicative',
        interval_width=0.95
    )
    
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        model.fit(prophet_data)
    
    # Forecast
    future = model.make_future_dataframe(periods=periods, freq='MS')
    forecast = model.predict(future)
    
    # Extract forecasts
    feb = forecast[forecast['ds'] == '2026-02-01'].iloc[0]
    mar = forecast[forecast['ds'] == '2026-03-01'].iloc[0]
    apr = forecast[forecast['ds'] == '2026-04-01'].iloc[0]
    
    return {
        'feb_forecast': feb['yhat'],
        'feb_lower': feb['yhat_lower'],
        'feb_upper': feb['yhat_upper'],
        'mar_forecast': mar['yhat'],
        'mar_lower': mar['yhat_lower'],
        'mar_upper': mar['yhat_upper'],
        'apr_forecast': apr['yhat'],
        'apr_lower': apr['yhat_lower'],
        'apr_upper': apr['yhat_upper']
    }

print("🔮 Generating Prophet forecasts for all states...")
print("This may take 2-3 minutes...\n")

prophet_forecasts = []

for idx, state in enumerate(states, 1):
    if idx % 5 == 0:
        print(f"Progress: {idx}/{total_states} states...")
    
    result = forecast_prophet(monthly_data, state)
    
    # Get current risk
    current = monthly_data[monthly_data['state'] == state]['total_risk'].iloc[-1]
    
    prophet_forecasts.append({
        'state': state,
        'current_risk_dec2025': current,
        'prophet_feb_2026': result['feb_forecast'],
        'prophet_feb_lower': result['feb_lower'],
        'prophet_feb_upper': result['feb_upper'],
        'prophet_mar_2026': result['mar_forecast'],
        'prophet_mar_lower': result['mar_lower'],
        'prophet_mar_upper': result['mar_upper'],
        'prophet_apr_2026': result['apr_forecast'],
        'prophet_apr_lower': result['apr_lower'],
        'prophet_apr_upper': result['apr_upper']
    })

prophet_forecast_df = pd.DataFrame(prophet_forecasts)

print(f"\n✅ Prophet forecasts complete!")
print(f"   States forecasted: {len(prophet_forecast_df)}")
print(f"\nFirst few rows:")
print(prophet_forecast_df[['state', 'current_risk_dec2025', 'prophet_feb_2026']].head())
```

## 💾 Save Prophet Results

```python
# Save Prophet forecasts
prophet_forecast_df.to_csv('../outputs/reports/prophet_forecasts_2026.csv', index=False)
print("✅ Saved: prophet_forecasts_2026.csv")

# Save validation results
prophet_val_df.to_csv('../outputs/reports/prophet_validation_results.csv', index=False)
print("✅ Saved: prophet_validation_results.csv")

# Create comparison table
comparison = pd.DataFrame({
    'model': ['Moving Average', 'Linear Trend', 'Prophet'],
    'rmse': [ma_rmse, lt_rmse, prophet_rmse],
    'improvement_vs_baseline': [
        0,
        ((ma_rmse - lt_rmse) / ma_rmse) * 100,
        ((ma_rmse - prophet_rmse) / ma_rmse) * 100
    ]
})

comparison.to_csv('../outputs/reports/final_model_comparison.csv', index=False)
print("✅ Saved: final_model_comparison.csv")
```

## 📊 Final Comparison Visualization

```python
# Model comparison bar chart
fig, ax = plt.subplots(figsize=(10, 6), dpi=100)

models = ['Moving Average', 'Linear Trend', 'Prophet']
errors = [ma_rmse, lt_rmse, prophet_rmse]
colors = ['orange' if e == min(errors) else 'lightblue' for e in errors]
colors[errors.index(min(errors))] = 'green'

bars = ax.bar(models, errors, color=colors, alpha=0.8, edgecolor='black', linewidth=1.5)

# Add value labels
for bar, error in zip(bars, errors):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{error:.2f}',
            ha='center', va='bottom', fontsize=12, fontweight='bold')

ax.set_ylabel('RMSE (Lower is Better)', fontsize=12, fontweight='bold')
ax.set_title('Final Model Comparison: Validation Performance', fontsize=14, fontweight='bold')
ax.set_ylim(0, max(errors) * 1.3)
ax.grid(True, alpha=0.3, axis='y')

# Add best model annotation
best_idx = errors.index(min(errors))
ax.annotate('BEST MODEL', 
            xy=(best_idx, errors[best_idx]), 
            xytext=(best_idx, errors[best_idx] + max(errors)*0.15),
            ha='center',
            fontsize=12,
            fontweight='bold',
            color='green',
            arrowprops=dict(arrowstyle='->', color='green', lw=2))

plt.tight_layout()
plt.savefig('../outputs/figures/week3_final_model_comparison.png', dpi=150, bbox_inches='tight')
print("✅ Saved: week3_final_model_comparison.png")
plt.show()
plt.close()
```

## 📈 Compare Forecasts: Prophet vs Linear Trend

```python
# Merge Prophet and simple forecasts
merged = simple_forecasts.merge(
    prophet_forecast_df[['state', 'prophet_feb_2026', 'prophet_feb_lower', 'prophet_feb_upper']],
    on='state'
)

# Calculate differences
merged['forecast_difference'] = merged['prophet_feb_2026'] - merged['best_forecast_feb']

print("📊 Forecast Comparison (Feb 2026):")
print("="*70)
print(f"Mean Simple Forecast: {merged['best_forecast_feb'].mean():.2f}")
print(f"Mean Prophet Forecast: {merged['prophet_feb_2026'].mean():.2f}")
print(f"Average Difference: {merged['forecast_difference'].mean():.2f}")
print("\nTop 5 states with largest differences:")
print(merged.nlargest(5, 'forecast_difference')[['state', 'best_forecast_feb', 'prophet_feb_2026', 'forecast_difference']])
```
